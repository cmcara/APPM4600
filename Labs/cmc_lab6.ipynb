```python
import numpy as np
import math
import time
from numpy.linalg import inv 
from numpy.linalg import norm 
```

```python
def evalF(x): 
    F = np.zeros(2)
    
    F[0] = x[0]**2 + x[1]**2 - 2
    F[1] = math.exp(x[0] - 1) + x[1]**2 - 2
 
    return F
    
def evalJ(x): 
    
    J = np.array([
        [2.0 * x[0],             2.0 * x[1]], 
        [math.exp(x[0] - 1),     2.0 * x[1]]
    ])

    return J
```

```python
def newton(x0,tol,Nmax):

    ''' inputs: x0 = initial guess, tol = tolerance, Nmax = max its'''
    ''' Outputs: xstar= approx root, ier = error message, its = num its'''

    for its in range(Nmax):
       J = evalJ(x0)
       Jinv = inv(J)
       F = evalF(x0)
       
       x1 = x0 - Jinv.dot(F)
       
       if (norm(x1-x0) < tol):
           xstar = x1
           ier =0
           return[xstar, ier, its]
           
       x0 = x1
    
    xstar = x1
    ier = 1
    return[xstar,ier,its]
```

```python
def lazynewton(x0,tol,Nmax):

    ''' Lazy Newton = use only the inverse of the Jacobian for initial guess'''
    ''' inputs: x0 = initial guess, tol = tolerance, Nmax = max its'''
    ''' Outputs: xstar= approx root, ier = error message, its = num its'''

    J = evalJ(x0)
    Jinv = inv(J)
    for its in range(Nmax):

       F = evalF(x0)
       x1 = x0 - Jinv.dot(F)
       
       if (norm(x1-x0) < tol):
           xstar = x1
           ier =0
           return[xstar, ier,its]
           
       x0 = x1
    
    xstar = x1
    ier = 1
    return[xstar,ier,its]   
```

```python
root, er, iters = newton([2,0.5], 10e-6, 10000)

print(f"{root}, {er}")
```

```python
root, er, iters = newton([3,5], 10e-6, 10000)

print(f"{root}, {er}")
```

```python
root, er, iters = lazynewton([2,0.5], 10e-6, 10000)

print(f"{root}, {er}")
```

```python
root, er, iters = lazynewton([3,5], 10e-6, 10000)

print(f"{root}, {er}")
```
