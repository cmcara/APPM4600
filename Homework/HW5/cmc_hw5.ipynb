> (a)

$ \begin{bmatrix} x_{n+1} \\ y_{n+1} \end{bmatrix} = \begin{bmatrix} x_n \\ y_n \end{bmatrix} - \begin{bmatrix} \frac{1}{6} & \frac{1}{18} \\ 0 & \frac{1}{6} \end{bmatrix} \begin{bmatrix} f(x_n,y_n) \\ g(x_n, y_n) \end{bmatrix}, n = 0, 1, 2, ...$

```python
def equations(x,y):
    f = 3 * x**2 - y**2 
    g = 3 * x * y**2 - x**3 - 1
    return np.array([f,g], dtype = float)

M = np.array([[1/6, 1/18],[0, 1/16]], dtype=float)

tol = 10e-5
max_iter = 25
```

```python
cur = np.array([1, 1])

print (f"Step 0: x = {cur[0]}, y = {cur[1]}")

for n in range(max_iter):
    
    F_val = equations(cur[0], cur[1])

    next_val = cur - M @ F_val

    change = np.linalg.norm(next_val - cur)

    cur = next_val

    print (f"Step {n+1}: x = {cur[0]}, y = {cur[1]} with change {change}")

    if change < tol:
        print (f"Converged on step {n+1} to x = {cur[0]}, y = {cur[1]}")
        break
```

This appears to converge but does so linearly, which is not very quick.

> (b)

The Jacobian of the system evaluated at (1,1) is $\begin{bmatrix} 6x & -2y \\ 3y^2 - 3x^2 & 6x \end{bmatrix} = \begin{bmatrix} 6(1) & -2(1) \\ 3(1)^2 - 3(1)^2 & 6(1) \end{bmatrix} = \begin{bmatrix} 6 & -2 \\ 0 & 6 \end{bmatrix}$ gives determinant = 6(6) - (-2)(0) = 36

which gives $J(1,1)^{-1}=\frac{1}{36} \begin{bmatrix} 6 & 2 \\ 0 & 6 \end{bmatrix} = \begin{bmatrix} 1/6 & 1/18 \\ 0 & 1/6 \end{bmatrix}$

The numerical matrix in the iteration uses the inverse of the jacobian at the initial guess to avoid recomputing the matrix inverse at each step.

>(c)

```python
def evalF(x): 

    F = np.zeros(2)
    
    F[0] = 3 * x[0]**2 - x[1]**2
    F[1] = 3 * x[0] * x[1]**2 - x[0]**3 - 1
 
    return F
    
def evalJ(x): 
    
    J = np.array([[6*x[0], -2*x[1]], 
        [3*x[1]**2 - 3*x[0]**2, 6*x[0]*x[1]]])
    
    return J

def Newton(x0,tol,Nmax):

    ''' inputs: x0 = initial guess, tol = tolerance, Nmax = max its'''
    ''' Outputs: xstar= approx root, ier = error message, its = num its'''

    for its in range(Nmax):
       J = evalJ(x0)
       Jinv = inv(J)
       F = evalF(x0)
       
       x1 = x0 - Jinv.dot(F)
       
       if (norm(x1-x0) < tol):
           xstar = x1
           ier =0
           return[xstar, ier, its]
           
       x0 = x1
    
    xstar = x1
    ier = 1
    return[xstar,ier,its]
```

```python
root, err, iters = Newton([1,1], tol, max_iter)

print(f"root: {root}, error: {err}, iterations: {iters}")
```

This converges much more quickly in terms of iteration count.

> # 2.
Given 
$\begin{cases} g_1 = \frac{1}{\sqrt{2}} \sqrt{1+(x+y)^2} - \frac{2}{3} \\ g_2 = \frac{1}{\sqrt{2}} \sqrt{1 + (x-y)^2} - \frac{2}{3} \end{cases}$


$|\frac{\partial g_i(x)}{\partial x_j}| \le \frac{K}{n}$

$x^{(k)} = G(x^{(k-1)})$, for each $k \ge 1$ converges to the unique fixed point $p \in D$ and $||x^{(k)}-p||_{\infty} \le \frac{K^k}{1-K}||x^{(1)}-x^{(0)}||_{\infty}$


Find  a region D in the x, y−plane for which the fixed point iteration $\begin{cases} x_{n+1} = \frac{1}{\sqrt{2}} \sqrt{1+(x_n+y_n)^2} - \frac{2}{3} \\ y_{n+1} = \frac{1}{\sqrt{2}} \sqrt{1 + (x_n-y_n)^2} - \frac{2}{3} \end{cases}$ is guaranteed to converge to a unique solution for any starting point $(x_0, y_0) \in D$.

We first show the sum of the partial derivatives is less than some constant K < 1.

$\frac{\partial g_1}{\partial x} = \frac{x+y}{\sqrt{2} \sqrt{1+(x+y)^2}}, \frac{\partial g_1}{\partial y} = \frac{x+y}{\sqrt{2} \sqrt{1+(x+y)^2}}$, gives $\frac{\partial g_1}{\partial x} + \frac{\partial g_1}{\partial y} = \sqrt{2} \frac{x+y}{\sqrt{1+(x+y)^2}}$

$\frac{\partial g_2}{\partial x} = \frac{x-y}{\sqrt{2} \sqrt{1+(x-y)^2}}, \frac{\partial g_2}{\partial y} = \frac{x-y}{\sqrt{2} \sqrt{1+(x-y)^2}}$, gives $\frac{\partial g_2}{\partial x} + \frac{\partial g_2}{\partial y} = \sqrt{2} \frac{x-y}{\sqrt{1+(x-y)^2}}$

letting u = (x+y) and v = (x-y), $K/2 = \sqrt{2} \frac{u}{\sqrt{1+(u)^2}} \lt \frac{1}{2}$ and $K/2 = \sqrt{2} \frac{v}{\sqrt{1+(v)^2}} \lt \frac{1}{2}$ gives $|x+y| < 1$ and $|x-y| < 1$

For $D = [-R, R] \times [-R, R]$, the max of $x,y \in [-R, R]$ is $x+y=2R$ and $x-y=2R$

If we choose a value for R such that plugging the max value (2R) into $K = \sqrt{2} \frac{u}{\sqrt{1+(u)^2}} = \sqrt{2} \frac{2R}{\sqrt{1+(2R)^2}} \lt 1$

$ 2 \frac{(2R)^2}{1 + (2R)^2} \lt 1$ gives $\frac{(2R)^2}{1 + (2R)^2} \lt \frac{1}{2}$ gives $\frac{4R^2}{1 + 4R^2} \lt \frac{1}{2}$

Letting R = .1 gives $K = \sqrt{2} \frac{2(.1)}{\sqrt{1 + (2(.1))}^2} = \sqrt{2} \frac{.2}{\sqrt{1 + .2^2}} = \sqrt{2} \frac{.2}{\sqrt{1 + .04}} = \sqrt{2} \frac{.2}{\sqrt{1.04}} = 0.2773500981 \lt 1$

Then $D = [-.1, .1] \times [-.1, .1]$ will guarantee convergence.

# 3. 

$\begin{cases} x_{n+1} = x_n - df_x \\ y_{n+1} = y_n - df_y \end{cases}$

$d = \frac{f(x_n, y_n)}{(f_x^2 + f_y^2)}$

>(a)

To move from $(x_n, y_n)$ off curve to a point on the curve along the normal line, the displacement vector must be parallel to the gradient vector 

$\Delta f = [f_x \ f_y]^T$ gives $x_{n+1} - x_n = \lambda f_x$ and  $y_{n+1} - y_n = \lambda f_y$

Using a first order Taylor expansion around $(x_n, y_n)$ for $f(x_{n+1}, y_{n+1}) = 0$ gives

$f(x_{n+1}, y_{n+1}) = f(x_n, y_n) + f_x(x_{n+1}-x_n) + f_y(y_{n+1}-y_n) = 0$

$f(x_n, y_n) + f_x(\lambda f_x) + f_y(\lambda f_y) = f(x_n, y_n) + \lambda(f_x)^2 + \lambda(f_y)^2 = f(x_n, y_n) + \lambda(f_x^2 + f_y^2) = 0$

$\lambda = - \frac{f(x_n, y_n)}{(f_x^2 + f_y^2)} = -d$

gives 

$x_{n+1} = x_n + \lambda f_x = x_n - df_x$ 


$y_{n+1} = y_n + \lambda f_y = y_n - df_y$


> (b)

```python
x, y, z = 1, 1, 1
max_iters = 20
tol = 10e-10

for i in range(max_iters):
    f_val = x**2 + 4 * y**2 + 4 * z**2 - 16

    fx = 2*x
    fy = 8*y
    fz = 8*z

    print(f"For step {i}: ({x}, {y}, {z}) gives {f_val}")

    if abs(f_val) < tol:
        print("Convergence")
        break

    d = f_val / (fx**2 + fy**2 + fz**2) 

    x = x - d * fx
    y = y - d * fy
    z = z - d * fz
```

This converges quadratically as the magnitude of the exponent of the residuals continues to roughly double each step.
