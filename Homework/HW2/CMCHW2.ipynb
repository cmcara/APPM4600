> (a)

Closed Interval [a,b] on which the equation has root r with IVT proof of r.

$f(x) = 2x - 1 - \sin{x}$ 

$f(0) = 2(0) - 1 - \sin{0} = 0 - 1 - 0 = -1$
$f(1) = 2(1) - 1 - \sin{1} = 2 - 1 - .841 = .159$

Since f(0) < 0 and f(1) > 0, there is a root r on the interval [0,1]. Since f(x) is continuos and the function is bounded by f(0) = -1 and f(1) = .159, there must be a root r such that f(r) = 0.

> (b)

the value of the derivative f'(x) = 2 - cos(x) will always stay above or equal to 1 for any x as cosign is bounded [-1,1], which means the function is strictly increasing for all x. A strictly increasing function can only cross y=0 once, which we identified above.

> (c)

```python
def bisect(a, b, func, tol):
    """
    f,a,b   - function and endpoints of initial interval
#   tol     - bisection stops when interval length < tol
    """

    # first verify there is a root we can find in the interval 

    fa = func(a)
    fb = func(b)

    if (fa*fb>0):
        ier = 1
        astar = a
        return [astar, ier, "Not in interval"]

    # verify end points are not a root 
    if (fa == 0):
        astar = a
        ier =0
        return [astar, ier, "A is a root"]

    if (fb ==0):
        astar = b
        ier = 0
        return [astar, ier, "B is a root"]

    count = 0
    d = .5 * (a + b)
    while (abs(d-a)> tol):
        fd = func(d)
        if (fd ==0):
            astar = d
            ier = 0
            return [astar, ier, count]
        if (fa*fd<0):
            b = d
        else: 
            a = d
            fa = fd
        d = .5 * (a+b)
        count = count +1
    astar = d
    ier = 0
    return [astar, ier, count]
```

```python
root, ier, iters = bisect(0,1,lambda x: 2*x - 1 - np.sin(x),1e-8)

if ier == 0:
    print("The root is:", root)
    print("The count is:", iters)
else:
    print("No root found")
```

$x_{n+1} = -\sin{(2x_n)} + \frac{5x_n}{4} - \frac{3}{4}$

```python
def fixedpt(x0, f,tol,Nmax):

    ''' x0 = initial guess''' 
    ''' Nmax = max number of iterations'''
    ''' tol = stopping tolerance'''

    count = 0
    v = []
    
    while (count <Nmax):
       count = count +1
       x1 = f(x0)
       v.append(x1)
       if (abs(x1-x0) <tol):
          xstar = x1
          ier = 0
          vec_full = np.array(v)
          return [xstar, ier, vec_full]
       vec_full = np.array(v)
       x0 = x1
    xstar = x1
    ier = 1
    return [xstar, ier, vec_full]
```

```python
root1, ier, vector = fixedpt(-1, lambda x: -np.sin(2*x) + (5*x)/4 - 3/4, 1e-9, 100)

if ier == 0:
    print("The root is:", root1)
else:
    print("No root found")
```
