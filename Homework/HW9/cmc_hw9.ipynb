$\phi_k(x) = x\phi_{k-1}(x) - b_k\phi_{k-1}(x) - c_k\phi_{k-2}(x) - \sum_{j=0}^{k-3} a_j \phi_j(x)$

Taking the inner product with a specific $\phi_j(x)$ for $j \leq k-3$

$<\phi_k, \phi_j>  =  <x\phi_{k-1},\phi_j> - b_k <\phi_{k-1},\phi_j> - c_k <\phi_{k-2},\phi_j> - a_j<\phi_j, \phi_j>$

$<\phi_k, \phi_j> = 0$ since $j \lt k$ and the definitely of orthogonal

$<x\phi_{k-1},\phi_j> = <\phi_{k-1},x\phi_j>$ where $x\phi_j$ has order $j+1 \lt k-1$ then $<x\phi_{k-1},\phi_j>=0$

$<\phi_{k-1},\phi_j> = 0$ and $<\phi_{k-2},\phi_j> = 0$ as well since $j \lt k-2 \lt k-1$

$\to 0 = 0 - b_k(0) - c_k(0) - a_j<\phi_j, \phi_j> \to a_j<\phi_j, \phi_j> = 0$ for all $a_j$ which forces $a_j = 0$ for all $j \leq k-3$

$a_0 = 0$

$a_1 = 1$

$b_3 = \frac{b_1}{6} \to \frac{b_1}{120} - \frac{b_1/6}{6} = \frac{b_1}{120} - \frac{b_1}{36} \to b_1(1/120 - 1/36) = 0 \to b_1 = 0$

$b_3 - \frac{b_1}{6} = 0 \to b_3 - 0 = 0 \to b_3 = 0$

$b_4 +\frac{1}{120} -\frac{1/6}{6} = 0 \to b_4 = -\frac{1}{120} +\frac{1}{36} = \frac{7}{360}$

$P_4^2 = \frac{0 + 1x + (0)x^2}{1+ 0x + \frac{1}{6}x^2 + 0x^3 + \frac{7}{360}x^4} = \frac{x}{1 + \frac{1}{6}x^2 + \frac{7}{360}x^4}$

```python
x = np.linspace(0, 5, 500)

#true
true_y = np.sin(x)

# 6th order maclaurin
m6 = x - (x ** 3) / 6 + (x ** 5) / 120

# (a) & (b)
pade_33 = (x - (7 / 60) * (x ** 3)) / (1 + (1 / 20) * (x ** 2))

# (c)
pade_24 = x / (1 + (1 / 6) * (x ** 2) + (7 / 360) * (x ** 4))

# calc errors
error_mac = np.abs(true_y - m6)
error_pade_33 = np.abs(true_y - pade_33)
error_pade_24 = np.abs(true_y - pade_24)

# plot
plt.plot(x, error_mac, label="6th-Order sinx Maclaurin", color="red")
plt.plot(x, error_pade_33, label="Padé [3/3]", color="blue")
plt.plot(x, error_pade_24, label="Padé [2/4]", color="green")

plt.yscale("log")  
plt.xlabel("x", fontsize=12)
plt.ylabel("Absolute Error (log scale)", fontsize=12)
plt.title(
    "Accuracy Comparison of sin(x) Approximations over [0, 5]", fontsize=14
)
plt.legend(fontsize=11)
plt.grid(True, which="both", ls="--", alpha=0.5)

plt.show()
```

The error for the approximation and Maclaurin expansion are very similar over the interval, which maxes sense as all three errors are bounded by $O(x^7)$. The $P^3_3$ approximation seems to start showing a clearer advantage near the end of the interval but the gap is still small.
